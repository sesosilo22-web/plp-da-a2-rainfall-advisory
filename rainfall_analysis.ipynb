#rainfall analysis

The rainfall analysis
Nairobi, Nakuru and Kisumu experienced their wetest month in april,Mombas in may, Machakos in november .
april and may be the the best consecutive planting months for Nirobi,Nakuru,Kisumu and Mombasasa,while November and December were the best for Machakos.
The NGO should consider these differences when planning seed distribution, planting activities and agricultural support in each county. However, Machakos had two missing rainfall readings, for April and September, which were replaced with the average of its valid readings (68 mm).
Therefore, the results should be treated as estimates, and the NGO should use additional historical rainfall records and weather forecasts before making final planting recommendations.
during the april rain the NGO could take their chance and plant after april rain since the soil is sufficiantly wet, but they should also consider the crop type ,soil conditions, drainage and local weather forecast


```python
import numpy as np
import csv
from google.colab import files

counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

clean_rows = [

[64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],  # Nairobi

[35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38], # Nakuru

[58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95], # Kisumu

[30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],  # Mombasa

]

months = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]

machakos_raw = ["48", "35", "88", "N/A", "140", " 20 ",

"9", "12", "", "45", "182", "96"]

valid,invalid =[],[]

#Cleaning machakos_raw using a function with try/except that converts valid values to integers.
#Replace invalid values with the average of the valid Machakos values (rounded), and print which months were imputed.
for v in machakos_raw:
  try:
    valid.append(int(v))
  except ValueError:
    invalid.append(v)

avg = round(sum(valid)/len(valid))
machakos_clean = []

for value in machakos_raw:
  try:
    machakos_clean.append(int(value))
  except ValueError:
    machakos_clean.append(avg)

    invalid_month =[]

    for month, value in zip(months, machakos_clean):
      if value == avg:
        invalid_month.append(month)
print('valid_values:', valid)
print('invalid_values:', invalid)
print('average:', avg)
print('clean_machakos:', machakos_clean)
print('imputed_months:', invalid_month)

#Combine all five counties into a single 5 × 12 NumPy array and confirm its shape and dtype.
rainfall= np.array( clean_rows + [machakos_clean])

print(rainfall)
print('data type:', rainfall.dtype)
print('shape:', rainfall.shape)

#Calculate annual total rainfall per county and the average rainfall per month across all counties.
print('total per countie:', rainfall.sum(axis=1))
print('average rainfall per month/:', rainfall.mean(axis=0))


#Using np.argmax with axis=1,
#find the wettest month for each county
#and print it by month name (use zip() from Section 2.9 to pair counties with results).

wettest_month = np.argmax(rainfall, axis=1)

print('wettest month for each county:', wettest_month)
for county, month in zip(counties, wettest_month):
  print(county, ':', months[month])



#Count, for each county, the number of “dry months” (below 50 mm) using a boolean mask.

dry_months = rainfall < 50
dry_count = np.sum(dry_months, axis=1)
print('dry months per county:', dry_count)

dry_mask = rainfall < 50

for i, county in enumerate(counties):
    dry_indices = np.where(dry_mask[i])[0]

    print(county)
    for index in dry_indices:
        print(index + 1, months[index], rainfall[i, index], "mm")
    print()


    #Identify the two consecutive months with the highest combined rainfall for each county;
    #this is the recommended planting window.
    #(Hint: the shifted-array technique R[:, :-1] + R[:, 1:] in Section 2.10.)

pairs = rainfall[:, :-1] + rainfall[:, 1:] #month pairs
best_pairs = pairs.max(axis=1)
print('planting window:', best_pairs)


for county, pair in zip(counties, best_pairs):

  print(county, ':', months[index], 'and',months[index+1])


#Write a CSV file rainfall_summary.csv with columns county, annual_total_mm,
#wettest_month, dry_months and planting_window.




annual_total_mm = rainfall.sum(axis=1)
wettest_month = np.argmax(rainfall, axis=1)
dry_months = np.sum(rainfall < 50, axis=1)
best_pairs = pairs.argmax(axis=1)

with open('rainfall_summary.csv', 'w', newline='') as f:
    writer = csv.DictWriter(
        f,
        fieldnames=[
            'county',
            'annual_total_mm',
            'wettest_month',
            'dry_months',
            'planting_window'

        ]

      )

    writer.writeheader()

    for i in range(len(counties)):
        writer.writerow({
            'county': counties[i],
            'annual_total_mm': annual_total_mm[i],
            'wettest_month': months[wettest_month[i]],
            'dry_months': dry_months[i],
            'planting_window': months[best_pairs[i]] + ' and ' + months[best_pairs[i] + 1]
        })
    files.download('rainfall_summary.csv')
    print('rainfall_summary.csv created')

```
